```python
from autoviz.AutoViz_Class import AutoViz_Class
import matplotlib.pyplot as plt

av = AutoViz_Class()

filename = "diabetes.csv"

df = av.AutoViz(
    filename=filename,
    sep=",",
    depVar="",          # leave empty for EDA
    dfte=None,
    header=0,
    verbose=1,
    lowess=False,
    chart_format="png",
    max_rows_analyzed=100000,
    max_cols_analyzed=30
)
```
